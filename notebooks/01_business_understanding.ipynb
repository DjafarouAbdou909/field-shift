## 1. Le problème

**Constat.** Dans les zones de savane ivoiriennes, le coton est la principale ressource économique, référence 1, page 325. Dans la région de Korhogo, il est passé de la culture associée à la culture pure, occupait environ 49 % des surfaces emblavées en 2002 et raccourcit la jachère, page 328. Les sols y sont peu humifères et de fertilité moyenne, et la saison des pluies va d'avril ou mai à octobre, page 327.

**Analyse.** L'agriculteur choisit chaque année ce qu'il sème sans lecture simple du climat de sa zone ni comparaison argumentée de rotations, ce qui reste une hypothèse à confirmer auprès de l'ANADER. Un début des pluies plus tardif est aussi une hypothèse, que les données doivent tester au lieu de la supposer.

**Décision.** L'outil compare au moins trois rotations, dont la rotation actuelle, à partir des données de la NASA, et ne parle d'un changement du climat que si les données l'établissent.

## 2. Les utilisateurs

**Constat.** L'agriculteur du Poro, producteur de coton et de maïs, est joint par WhatsApp en français, en texte et en voix. Son conseiller de l'ANADER, agence de conseil agricole et rural créée le 29 septembre 1993, référence 2, et le jury utilisent l'application web. La région est peuplée surtout de Sénoufo et de Dioula, référence 1, page 327.

**Analyse.** Que coton puis maïs soit la rotation la plus courante, que l'agriculteur utilise WhatsApp et lise le français sont des hypothèses à confirmer.

**Décision.** La voix accompagne le texte, et l'application détaille chaque chiffre pour le conseiller.

## 3. Les décisions éclairées et ce que l'outil ne fera pas

**Constat.** L'outil éclaire quatre décisions, garder sa rotation ou en discuter une autre, introduire une légumineuse, remplacer une partie du maïs par une céréale plus sobre, et pondérer ces choix selon sa priorité, l'eau, le revenu ou le sol.

**Analyse.** Le climat de saison n'explique que 0,9 % à 6,3 % de la variation annuelle des rendements dans le projet antérieur, référence 4. Une prédiction du rendement de chaque rotation ne serait donc pas défendable.

**Décision.** L'outil ne prédit pas de rendement, ne désigne pas de meilleure rotation prouvée, ne remplace ni l'agriculteur ni son conseiller, montre la date habituelle de début des pluies sans conseiller de date de semis, ne donne ni dose d'engrais ni traitement, ne couvre que Korhogo et Bouaké, ne traite que la première saison des pluies de Bouaké et ne parle que le français.

## 4. Les critères de réussite

**Constat.** Le cadrage fixe sept critères.

1. Au moins trois stratégies par ville, dont la rotation actuelle, chaque élément classé `mesure`, `calcule` ou `hypothese`.
2. Chaque chiffre reproductible par une commande, avec sa plage interannuelle et sa source.
3. Des vérifications de bon sens, dont un écart entre Korhogo et Bouaké qui suit les données ou dont l'absence est expliquée.
4. Le scénario de sensibilité qui change le classement, s'il y en a un, nommé.
5. Un parcours WhatsApp complet sur données réelles, sous le garde-fou.
6. Une relecture par un agronome ou un conseiller ANADER, ou son absence écrite.
7. Chaque règle de calcul explicable par chaque membre de l'équipe.

**Analyse.** Aucun de ces critères ne mesure une précision, faute de vérité terrain sur la meilleure rotation.

**Décision.** La phase 5 vérifiera chaque critère et écrira ce qui n'est pas atteint.
